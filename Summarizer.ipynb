```python
import spacy
from transformers import pipeline
```

```python
nlp = spacy.load("en_core_web_trf")
```

```python
summarizer = pipeline("summarization", model="facebook/bart-large-cnn", device = "cpu")
```

```python
def preprocess_text(text):
    doc = nlp(text)
    sentences = [sent.text.strip() for sent in doc.sents if sent.text.strip()]
    return " ".join(sentences)
```

```python
def summarize(text, max_length=100, min_length=30):
    summary = summarizer(text, max_length=max_length, min_length=min_length, do_sample=False)
    return summary[0]['summary_text']
```

```python
raw_text = '''Rome began as a small settlement in central Italy, traditionally founded in 753 BCE. It grew into a kingdom, then a republic (509 BCE), marked by expansion and internal strife. Julius Caesar's rise led to the fall of the Republic and the rise of the Roman Empire in 27 BCE under Augustus. The Empire expanded across Europe, Africa, and Asia, flourishing culturally and militarily. Internal decay and invasions led to its fall in the West in 476 CE, while the Eastern Roman (Byzantine) Empire endured until 1453. Rome remained influential, especially through the Catholic Church, which made it a spiritual capital.'''
print("Original Length:", len(raw_text))
```

```python
clean_text = preprocess_text(raw_text)
```

```python
summary = summarize(clean_text)
```

```python
print("Final Summary:\n", summary)
print("\nLength of final summary:", len(summary))
```
